# 🎥 Render the Lean course videos on Colab (Remotion)

Renders the animated lessons in the cloud so you don't depend on your PC.
No GPU needed (CPU is fine). Run the cells top to bottom.

**Compositions you can render:**
- `EightWastes` — short lesson (~73s) with captions
- `EightWastesDub` — same, no burned captions (for dubbing)
- `FullEightWastes` — full ~9 min lesson
- `CharacterSample` — the character demo

```python
# 1) Install Node.js 20
!curl -fsSL https://deb.nodesource.com/setup_20.x | bash - >/dev/null 2>&1
!apt-get install -y nodejs >/dev/null 2>&1
!node -v && npm -v
```

```python
# 2) Install the Chromium libraries Remotion needs for headless rendering
!apt-get update -qq >/dev/null 2>&1
!apt-get install -y -qq libnss3 libatk1.0-0 libatk-bridge2.0-0 libcups2 libdrm2 \
  libxkbcommon0 libxcomposite1 libxdamage1 libxfixes3 libxrandr2 libgbm1 libasound2 \
  libpango-1.0-0 libcairo2 fonts-liberation >/dev/null 2>&1
print('chromium libs installed')
```

```python
# 3) Get the project.
# PUBLIC repo: this works as-is. PRIVATE repo: put a GitHub token in TOKEN.
TOKEN = ""  # e.g. "ghp_xxx"  (leave empty if the repo is public)
import os
repo = "github.com/PrashantRMarathe/voice-replacer.git"
url = f"https://{TOKEN + '@' if TOKEN else ''}{repo}"
!rm -rf voice-replacer
!git clone -b add-course-video {url}
%cd voice-replacer/course_video
print('cwd:', os.getcwd())
```

```python
# 4) Install dependencies (~2-3 min)
!npm install
```

```python
# 5) Render. Change COMP to the composition you want.
COMP = "EightWastes"   # or EightWastesDub / FullEightWastes / CharacterSample
OUT  = "out/lesson.mp4"
# --gl=angle works on most Colab VMs; if it errors, change to swangle.
!npx remotion render {COMP} {OUT} --gl=angle --concurrency=2
```

```python
# 6) Download the rendered video
from google.colab import files
files.download('out/lesson.mp4')
```
